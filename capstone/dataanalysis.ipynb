```python
import seaborn as sns
import numpy as np

import matplotlib.pyplot as plt

# Basic overview
print("Shape:", df.shape)
print("\nDtypes:")
print(df.dtypes)
print("\nTop missing counts:")
print(df.isna().sum().sort_values(ascending=False).head(20))

# Quick value counts for key columns (if present)
for col in ['asset_type', 'substation_region', 'grid_cluster_id', 'grid_failure_flag']:
    if col in df.columns:
        print(f"\nValue counts for {col}:")
        print(df[col].value_counts(dropna=False).head(20))

# Normalize/derive failure binary flag
if 'grid_failure_flag' in df.columns:
    def to_bin(x):
        if pd.isna(x): 
            return 0
        if isinstance(x, (int, float)):
            return 1 if x == 1 else 0
        s = str(x).strip().lower()
        return 1 if s in ('1','true','t','y','yes') else 0
    df['_grid_failure_bin'] = df['grid_failure_flag'].apply(to_bin)
    if 'asset_type' in df.columns:
        print("\nFailure rate by asset_type (proportion of rows flagged):")
        print(df.groupby('asset_type')['_grid_failure_bin'].mean().sort_values(ascending=False))

# Normalize failure causes using domain-driven categories (weather, vegetation, animals, overload, aging, equipment, human/third-party, manufacturing, other)
possible_cause_cols = [c for c in df.columns if c.lower() in ('failure_cause','cause','outage_cause','root_cause')]
if possible_cause_cols:
    cause_col = possible_cause_cols[0]
    print(f"\nUsing cause column: {cause_col}")
    def normalize_cause(s):
        if pd.isna(s): return 'Unknown'
        s = str(s).lower()
        if any(k in s for k in ['weather','storm','wind','lightning','hail','ice','flood']):
            return 'Weather'
        if any(k in s for k in ['veget','tree','overgrown','brush']):
            return 'Vegetation'
        if any(k in s for k in ['animal','bird','rodent']):
            return 'Animal'
        if any(k in s for k in ['overload','over current','over-current','overcurrent','over load']):
            return 'Overload'
        if any(k in s for k in ['aging','age','corrosion','wear','deterioration']):
            return 'Aging/Degradation'
        if any(k in s for k in ['manufactur','defect','faulty','design']):
            return 'Manufacturing/Design'
        if any(k in s for k in ['human','digging','third party','vandal','excavation','construction']):
            return 'Human/Third-party'
        if any(k in s for k in ['equipment','transformer','breaker','cable','insulator','switchgear']):
            return 'Equipment'
        return 'Other'
    df['cause_norm'] = df[cause_col].apply(normalize_cause)
    print("\nNormalized cause counts:")
    print(df['cause_norm'].value_counts())
    if '_grid_failure_bin' in df.columns:
        print("\nFailure rate by normalized cause:")
        print(df.groupby('cause_norm')['_grid_failure_bin'].mean().sort_values(ascending=False))

    # Plot normalized causes
    plt.figure(figsize=(8,4))
    sns.countplot(data=df, y='cause_norm', order=df['cause_norm'].value_counts().index, palette='tab10')
    plt.title('Normalized Failure Causes')
    plt.tight_layout()
    plt.show()

# Time-series of failures if a date-like column exists
date_candidates = [c for c in df.columns if c.lower() in ('failure_date','date','event_date','outage_date','reported_date','timestamp')]
if date_candidates:
    date_col = date_candidates[0]
    print(f"\nUsing date column: {date_col}")
    df['_dt'] = pd.to_datetime(df[date_col], errors='coerce')
    if df['_dt'].notna().any():
        monthly = df.set_index('_dt').resample('M').size()
        plt.figure(figsize=(10,3))
        monthly.plot()
        plt.title('Events per Month')
        plt.ylabel('Count')
        plt.tight_layout()
        plt.show()
        if '_grid_failure_bin' in df.columns:
            monthly_failures = df[df['_grid_failure_bin']==1].set_index('_dt').resample('M').size()
            plt.figure(figsize=(10,3))
            monthly_failures.plot(color='r')
            plt.title('Failures per Month')
            plt.ylabel('Failure Count')
            plt.tight_layout()
            plt.show()

# Numeric correlations (to spot predictors of failure duration/severity if numeric columns exist)
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
if len(num_cols) > 1:
    corr = df[num_cols].corr()
    plt.figure(figsize=(8,6))
    sns.heatmap(corr, annot=True, fmt=".2f", cmap='vlag', center=0)
    plt.title('Numeric Correlation Matrix')
    plt.tight_layout()
    plt.show()

# Age / asset condition exploration if present
age_candidates = [c for c in df.columns if 'age' in c.lower() or 'years' in c.lower() or 'asset_age' in c.lower()]
if age_candidates:
    age_col = age_candidates[0]
    print(f"\nUsing age column: {age_col}")
    plt.figure(figsize=(6,3))
    sns.histplot(df[age_col].dropna(), bins=30, kde=True)
    plt.title('Asset Age Distribution')
    plt.tight_layout()
    plt.show()
    if '_grid_failure_bin' in df.columns:
        print("\nMean age by failure flag:")
        print(df.groupby('_grid_failure_bin')[age_col].mean())

# Suggest next steps (printed as short checklist)
print("\nNext analytical steps you may want to run:\n- Drill failure rates by (asset_type, substation_region, grid_cluster_id)\n- Inspect outage_duration / time-to-repair vs cause\n- Map causes to preventive actions (vegetation management, inspections for aging assets, lightning protection, animal guards)")
```

```python
# Clean duplicates and non-meaningful rows
import pandas as pd

df = pd.read_csv('PowerGrid_Utility_Intelligence.csv')
df['asset_type'] = df['asset_type'].replace({'TRANSFORMER': 'Transformer', 'transformer': 'Transformer'})
num_before = len(df)

# 1) remove exact duplicate rows
df = df.drop_duplicates()
num_after = len(df)
print(f"Rows before: {num_before}, after dropping duplicates: {num_after}, removed: {num_before - num_after}")

# 2) if there are multiple records per asset, keep first by asset_id
if 'asset_id' in df.columns:
    df = df.drop_duplicates(subset='asset_id', keep='first')

num_after = len(df)
print(f"Rows before: {num_before}, after dropping multiple records per asset: {num_after}, removed: {num_before - num_after}")

# 3) drop rows missing core identifiers
df = df.dropna(subset=[c for c in ['asset_id','asset_type','substation_region'] if c in df.columns])

num_after = len(df)
print(f"Rows before: {num_before}, after dropping na: {num_after}, removed: {num_before - num_after}")

# 4) remove obviously invalid asset_type values
if 'asset_type' in df.columns:
    bad_types = {'', 'none', 'nan', 'unknown', 'n/a', 'na'}
    df = df[~df['asset_type'].astype(str).str.strip().str.lower().isin(bad_types)]

num_after = len(df)
print(f"Rows before: {num_before}, after dropping invalid asset type: {num_after}, removed: {num_before - num_after}")


# 5) remove rows with impossible ages (if column exists)
if 'asset_age_years' in df.columns:
    df = df[df['asset_age_years'].notna() & (df['asset_age_years'] >= 0)]

num_after = len(df)
print(f"Rows before: {num_before}, after dropping asset age in years: {num_after}, removed: {num_before - num_after}")


# 6) drop rows that are entirely NA (defensive)
df = df.dropna(how='all')
num_after = len(df)
print(f"Rows before: {num_before}, after dropping all na column rows: {num_after}, removed: {num_before - num_after}")


# reset index and report
df = df.reset_index(drop=True)
num_after = len(df)
print(f"Rows before: {num_before}, after cleaning: {num_after}, removed: {num_before - num_after}")

missing = df.isnull().sum().sort_values(ascending=False)
print(missing)

oil_missing_pct = (
    df.groupby('asset_type')['oil_quality_score']
      .apply(lambda x: x.isnull().mean()*100)
)

print(oil_missing_pct)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
df['oil_quality_score'].hist(bins=30)
plt.title('Oil Quality Score')

plt.subplot(1, 2, 2)
df['inspection_score'].hist(bins=30)
plt.title('Inspection Score')

plt.show()
```

```python
oil_missing = (
    df.groupby('grid_failure_flag')['oil_quality_score']
      .apply(lambda x: x.isnull().mean() * 100)
)

print(oil_missing)

inspection_missing = (
    df.groupby('grid_failure_flag')['inspection_score']
      .apply(lambda x: x.isnull().mean() * 100)
)

print(inspection_missing)

# median imputation for oil_quality_score and inspection_score - industry standard practice for missing values in predictive modeling
df['oil_quality_score'] = df['oil_quality_score'].fillna(
    df['oil_quality_score'].median()
)

df['inspection_score'] = df['inspection_score'].fillna(
    df['inspection_score'].median()
)
```
