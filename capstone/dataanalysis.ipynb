```python
import pandas as pd

df = pd.read_csv('PowerGrid_Utility_Intelligence.csv')
df.head()

```

```python
df.groupby('asset_type').size()
```

```python
# clean the data - change TRANSFORMER and transformer to Transformer in the 'asset_type' column.
df['asset_type'] = df['asset_type'].replace({'TRANSFORMER': 'Transformer', 'transformer': 'Transformer'})
df.groupby('asset_type').size()

df.groupby(['substation_region', 'asset_type']).size()
```

```python
# show all column data for asset_type = 'Transformer' and substation_region = 'North'
df[(df['asset_type'] == 'Transformer') & (df['substation_region'] == 'North')]
```

```python
df[(df['asset_type'] == 'Substation') & (df['substation_region'] == 'North')]
```

```python
df.sort_values(['substation_region', 'asset_type']).groupby(['substation_region', 'asset_type']).head(1)
```

```python
# group by grid_cluster_id, substation_region, asset_type and get the count of each group
df.groupby(['grid_cluster_id', 'substation_region', 'asset_type']).size()
```

```python
# Add grid_cluster_id, substation_region and asset_type to the breakdown
df.groupby(['grid_failure_flag', 'grid_cluster_id', 'substation_region', 'asset_type']).size()
```

```python
df[(df['substation_region'] == 'Central') & (df['grid_cluster_id'] == 1)].groupby(['grid_failure_flag', 'grid_cluster_id', 'substation_region', 'asset_type']).size()
```
